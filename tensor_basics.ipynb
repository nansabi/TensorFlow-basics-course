```python
import tensorflow as tf
```

```python
scalar = tf.constant(5)
scalar
```

```python
vector = tf.constant([1,2,3])
vector
```

```python
matrix = tf.constant([[1,2,3],[4,5,6]])
matrix
```

```python
tensor = tf.constant([[[1,2,3,4],[5,6,7,8]],[[3,7,4,7],[9,7,4,6]]])
tensor
```

```python
print(scalar.shape , tf.rank(scalar).numpy())
print(tensor.shape , tf.rank(tensor).numpy())
print(matrix.shape , tf.rank(matrix).numpy())
```

```python
a = tf.constant([1,2,3])
b = tf.constant([3,4,5])

print(a + b)
print((a+b).numpy())
print(a-b)
print(a*b)
print(a/b)
```
